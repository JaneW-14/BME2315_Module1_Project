## Project Goal:


Neuroscientists have found that learning can physically alter the brain. A study by Asano, Taki, Hashizume et al. (2014) conducted a longitudinal study of 229 healthy children in Japan between the ages of 5 and 18. They aimed to see whether the amount of time children studied at home was related to the amount of white and gray matter. They did an MRI scan of every participant at the beginning of the study and one at the end, 3 years later. They found a positive correlation between studying 1+ hour a night and increased white matter in the right precuneus and right superior temporal gyrus. Furthermore, they found a positive correlation with change in gray matter in the right precuneus and the right inferior parietal lobe (Asano, Taki, Hashizume et al. 2014). Other neuroscientists have found that the right superior frontal gyrus is known to play an important role in memory. Furthermore, increased white and gray matter is known to be a sign of an increased number of axons. Therefore, we hypothesize that if learning increases the number of axons in the areas of the brain that are important for memory, the memory loss symptom of Alzieherms (dementia) will be delayed, and the age at onset will be larger.
This article inspired us to investigate whether the number of years of education a patient had impacted the age at which their Alzheimer's symptoms started. This can be determined by comparing the years spent in education and age of onset for each patient. We are also curious if this relationship would change depending on the patient’s genetic predisposition. We can investigate this by creating subgroups filtered by the patient’s version of the apolipoprotein E (APOE) gene. The 3 variants of APOE are Ɛ2, Ɛ3, and Ɛ4. Every human has two of these variants, making 6 possible combinations: 2/2, 2/3, 2/4, 3/3, 3/4, and 4/4. Individuals with an Ɛ4 allele have a higher risk of getting AD, and patients with two Ɛ4 alleles (4/4) are at the highest risk.

### Disease progression & prognosis


Alzheimer’s disease progresses differently for each person. On average, people live about **8 years after symptoms begin**, though some live for **20 years or longer**. Early-onset Alzheimer’s is rare and can begin as early as a person’s 30s or 40s.

1. **Preclinical stage:** Brain changes begin before noticeable symptoms appear.
2. **Early stage:** Mild memory and thinking problems may affect daily activities.

### Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)



Alzheimer’s disease causes changes in the brain that damage neurons and disrupt their connections. Two key changes are:

For our question we compared **APOE genotype** (categories 2_3, 3_3, 3_4, and 4_4), **years of education** (in years), and **age of onset of cognitive symptoms** (in years). ## Conclusions and Ethical Implications: 
The first research question investigated whether the number of years an individual was educated had an impact on the age of onset of symptoms. We analyzed the data using an ANOVA test, which resulted in a p-value = 0.278, which is not less than the commonly used significance value of 0.05. Therefore, the ANOVA test failed and we failed to reject the null hypothesis. There is no significant difference between the means of the different groups, and therefore, the age of onset symptoms is not impacted by the number of years of education. 

​

The second research question focused on the impact of years of education on the age of onset symptoms for each gene variant. The analysis method included graphing a scatterplot and applying a linear regression. Gene variants 4_4, 3_4, 3_3, and 2_4 were successfully graphed individually. None of the linear regressions resulted in a value of R2 close to one, which means there is no relationship between the number of years of education and the age of onset of symptoms for patients with any gene variant. 

​

There are limitations to our data and analysis and, therefore, inherently to our conclusions. Furthermore, across the other genes, the number of data points was way too small to draw an accurate conclusion. Therefore, the conclusions drawn between the number of years of education and age of onset within specific gene variants have very low validity. 

​

The data analysis for the first research question has a considerable number of data points. While increasing the number of data points added would increase the validity of the conclusion, the conclusion we drew is still worth examining. If there is no significant impact of years of education on the age of onset of symptoms, it can be argued that continual cognitive stimuli does not defer the symptoms of Alzheimer's. In this case, medical professionals and researchers should focus their attention on other possible mediators that could delay symptoms. The ethical implication is the possible consequences and successes resulting from redirecting researchers’ focus to treatments that have a higher chance of managing symptoms, which would require reducing the amount of research on the role of cognitive stimuli. 

## Limitations and Future Work: 
The biggest limitation in this data set, and therefore our analysis, is the small number of data points. As mentioned in the conclusion section, the lack of data for the individual gene variants nullifies the conclusions drawn from the linear regression model. 

​

For the data analysis of the entire data set, comparing the number of years of education and age of onset of symptoms, an increase in data points would improve the strength of the conclusion. Additionally, the number of years of education is incredibly vague. There is no way to know the type of education, the difficulty level, and therefore the level of cognitive stimuli the education provided. Without these details, it is very hard to generalize the findings to all forms of cognitive stimuli. In other words, it would not be surprising if there were studies that contradicted these findings because they had a more accurate way to measure cognitive stimuli. 

​

In the future, to further analyze the relationship between the number of years of education and the age of onset of symptoms, we would start by using a much bigger data set. We would then be very specific about what was considered cognitive stimuli. We would also adjust for cultural differences. We would need to know what each culture considered cognitive stimuli to accurately compare patient data between cultures. For example, if we were to find data from a study similar to Asano, Taki, Hashizume et al. 2014, we would have to be careful about analyzing the amount of studying each student does because there could be discrepancies about how intense studying should be across cultures.
